```python
# ============================================================
# CUSTOMER CHURN PREDICTION - EXPLORATORY DATA ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
import io

sns.set_theme(style="whitegrid")

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

uploaded = files.upload()
file_name = list(uploaded.keys())[0]

df = pd.read_csv(io.BytesIO(uploaded[file_name]))

print("Dataset loaded successfully.")
print("Shape:", df.shape)

# ------------------------------------------------------------
# 2. BASIC DATASET INFORMATION
# ------------------------------------------------------------

print("\nFIRST 5 ROWS")
display(df.head())

print("\nDATASET INFORMATION")
df.info()

print("\nSTATISTICAL SUMMARY")
display(df.describe().T)

print("\nCOLUMN NAMES")
print(list(df.columns))

# ------------------------------------------------------------
# 3. MISSING VALUES
# ------------------------------------------------------------

print("\nMISSING VALUE ANALYSIS")

# Check blank TotalCharges values before conversion
blank_totalcharges = (
    df["TotalCharges"].astype(str).str.strip().eq("").sum()
)

print("Blank TotalCharges values:", blank_totalcharges)

# Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

missing = df.isnull().sum()
missing = missing[missing > 0]

if len(missing) > 0:
    display(missing.to_frame("Missing Values"))

    plt.figure(figsize=(7, 4))
    plt.bar(missing.index, missing.values)
    plt.title("Missing Values by Column")
    plt.xlabel("Column")
    plt.ylabel("Number of Missing Values")
    plt.show()
else:
    print("No missing values found.")

# ------------------------------------------------------------
# 4. DUPLICATE ANALYSIS
# ------------------------------------------------------------

print("\nDUPLICATE ANALYSIS")
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:", df["customerID"].duplicated().sum())

# ------------------------------------------------------------
# 5. CHURN DISTRIBUTION
# ------------------------------------------------------------

print("\nCHURN DISTRIBUTION")

churn_counts = df["Churn"].value_counts()
churn_percent = df["Churn"].value_counts(normalize=True) * 100

display(
    pd.DataFrame({
        "Count": churn_counts,
        "Percentage": churn_percent.round(2)
    })
)

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()

# ------------------------------------------------------------
# 6. NUMERICAL FEATURES
# ------------------------------------------------------------

numerical_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

print("\nNUMERICAL FEATURE SUMMARY")
display(df[numerical_features].describe().T)

# ------------------------------------------------------------
# 7. NUMERICAL DISTRIBUTIONS
# ------------------------------------------------------------

for column in numerical_features:

    plt.figure(figsize=(7, 4))
    sns.histplot(data=df, x=column, kde=True)
    plt.title("Distribution of " + column)
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.show()

# ------------------------------------------------------------
# 8. CHURN VS TENURE
# ------------------------------------------------------------

print("\nAVERAGE TENURE BY CHURN")
display(
    df.groupby("Churn")["tenure"].mean().round(2)
)

plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="Churn", y="tenure")
plt.title("Tenure by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")
plt.show()

# ------------------------------------------------------------
# 9. CHURN VS MONTHLY CHARGES
# ------------------------------------------------------------

print("\nAVERAGE MONTHLY CHARGES BY CHURN")
display(
    df.groupby("Churn")["MonthlyCharges"].mean().round(2)
)

plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="Churn", y="MonthlyCharges")
plt.title("Monthly Charges by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.show()

# ------------------------------------------------------------
# 10. CHURN VS CONTRACT
# ------------------------------------------------------------

print("\nCHURN RATE BY CONTRACT")

contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

display(contract_churn.round(2))

plt.figure(figsize=(8, 4))
contract_churn["Yes"].plot(kind="bar")
plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.show()

# ------------------------------------------------------------
# 11. CHURN VS INTERNET SERVICE
# ------------------------------------------------------------

print("\nCHURN RATE BY INTERNET SERVICE")

internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

display(internet_churn.round(2))

plt.figure(figsize=(8, 4))
internet_churn["Yes"].plot(kind="bar")
plt.title("Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.show()

# ------------------------------------------------------------
# 12. CORRELATION ANALYSIS
# ------------------------------------------------------------

print("\nCORRELATION MATRIX")

correlation = df[numerical_features].corr()
display(correlation.round(3))

plt.figure(figsize=(7, 5))
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.show()

# ------------------------------------------------------------
# 13. EDA SUMMARY
# ------------------------------------------------------------

print("\nEDA SUMMARY")
print("""
1. The dataset contains 7,043 customer records and 21 columns.

2. The target variable Churn contains two classes: Yes and No.

3. The dataset shows class imbalance, with fewer customers
   belonging to the churn class.

4. TotalCharges contains 11 blank values, which are converted
   to missing values during preprocessing.

5. No duplicate rows or duplicate customer IDs were identified.

6. Churned customers have lower average tenure than
   non-churned customers.

7. Churned customers have higher average monthly charges.

8. Month-to-month contract customers show a substantially
   higher churn rate than one-year and two-year contract
   customers.

9. Internet service type also shows differences in churn rates.

10. The correlation analysis shows a strong relationship
    between tenure and TotalCharges.

11. These EDA findings support preprocessing, feature
    transformation, regularization and model development.
""")

print("\nEDA COMPLETED SUCCESSFULLY.")
```
